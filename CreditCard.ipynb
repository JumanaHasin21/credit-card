```python
# Function to apply IQR method to fix outliers
def fix_outliers_iqr(df):
    num=df.select_dtypes(include='number').columns
    for col in num:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
# Replace outliers with the respective bounds
        df[col] = df[col].apply(lambda x: lower_bound if x < lower_bound else upper_bound if x > upper_bound else x)
    return df
```

```python

# Identify the target and categorical columns
target_column = "default payment next month"

categorical_columns = [
    "SEX",
    "EDUCATION",
    "MARRIAGE"
]

print("Target column:", target_column)
print("Categorical columns:", categorical_columns)
```

```python

# Separate independent and dependent variables

X = df.drop(
    columns=[target_column, "ID"],
    errors="ignore"
)

y = df[target_column].astype(int)

print("Independent variables (X):", X.shape)
print("Dependent variable (y):", y.shape)

print("\nTarget classes:")
print(y.value_counts())
```

```python
# Rename columns for consistency

df = df.rename(columns={
    "default payment next month": "default_payment_next_month",
    "PAY_0": "PAY_1"
})

X = X.rename(columns={"PAY_0": "PAY_1"})
y.name = "default_payment_next_month"

print("Updated column names:")
print(df.columns.tolist())
```

```python
# Split the dataset into training and testing sets

from sklearn.model_selection import train_test_split

RANDOM_STATE = 42

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python
# Preprocessing: One-Hot Encoding and StandardScaler

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Identify numerical and categorical features
numeric_features = X_train.select_dtypes(include="number").columns.tolist()

categorical_features = [
    column for column in categorical_columns
    if column in X_train.columns
]

# Remove categorical columns from numerical features
numeric_features = [
    column for column in numeric_features
    if column not in categorical_features
]

# Numerical preprocessing
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing methods
preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_transformer, numeric_features),
        ("categorical", categorical_transformer, categorical_features)
    ],
    remainder="drop"
)

print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)
```

```python
# Feature selection using SelectKBest

from sklearn.feature_selection import SelectKBest, f_classif

# Fit preprocessing only on training data
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

# Get feature names after preprocessing
feature_names = preprocessor.get_feature_names_out()

# Select the 20 best features
selector = SelectKBest(score_func=f_classif, k=20)

X_train_selected = selector.fit_transform(
    X_train_processed, y_train
)

X_test_selected = selector.transform(X_test_processed)

# Display selected features
selected_features = feature_names[selector.get_support()]

print("Selected Features:")
print(selected_features)

print("\nTraining data after feature selection:", X_train_selected.shape)
print("Testing data after feature selection:", X_test_selected.shape)
```

```python
# Selecting classification algorithms

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

models = {
    "Logistic Regression": LogisticRegression(),
    "Decision Tree": DecisionTreeClassifier(random_state=RANDOM_STATE),
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=RANDOM_STATE
    )
}

print("Models selected:")
for name in models:
    print(name)
```

```python
# Train models and evaluate their performance

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

scores = []

for name, model in models.items():

    # Train the model
    model.fit(X_train_selected, y_train)

    # Make predictions
    y_pred = model.predict(X_test_selected)

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)

    scores.append({
        "Model": name,
        "Accuracy": round(accuracy, 4),
        "Precision": round(precision, 4),
        "Recall": round(recall, 4),
        "F1-Score": round(f1, 4)
    })

# Display the comparison
results_df = pd.DataFrame(scores)

print("Model Performance Comparison:")
display(results_df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

metrics = ["Accuracy", "Precision", "Recall", "F1-Score"]

results_plot = results_df.set_index("Model")[metrics]

results_plot.plot(
    kind="bar",
    figsize=(10, 6),
    width=0.8
)

plt.title("Model Performance Comparison")
plt.xlabel("Classification Models")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(title="Evaluation Metrics")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, model) in zip(axes, models.items()):
    y_pred = model.predict(X_test_selected)

    cm = confusion_matrix(y_test, y_pred)

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["No Default", "Default"]
    )

    disp.plot(ax=ax, cmap="Blues", colorbar=False)
    ax.set_title(name)

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import roc_auc_score

for name, model in models.items():
    y_prob = model.predict_proba(X_test_selected)[:, 1]
    auc = roc_auc_score(y_test, y_prob)

    print(f"{name}: ROC-AUC = {auc:.4f}")
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

for name, model in models.items():
    y_prob = model.predict_proba(X_test_selected)[:, 1]

    fpr, tpr, thresholds = roc_curve(y_test, y_prob)
    auc = roc_auc_score(y_test, y_prob)

    plt.plot(fpr, tpr, label=f"{name} (AUC = {auc:.4f})")

plt.plot([0, 1], [0, 1], linestyle="--", label="Random Classifier")

plt.title("ROC Curve Comparison")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.grid()
plt.show()
```

```python
## Conclusion
Three classification algorithms were implemented and evaluated to predict credit card payment defaults: Logistic Regression, Decision Tree, and Random Forest.

The models were compared using Accuracy, Precision, Recall, F1-Score, ROC-AUC, and Confusion Matrix.

- Logistic Regression achieved an accuracy of 80.63% and the highest precision of 68.05%.
- Decision Tree achieved the highest recall of 39.26%, identifying more actual defaulters.
- Random Forest achieved the highest accuracy of 81.10%, F1-Score of 46.56%, and ROC-AUC of 0.7436.

Based on the F1-Score, Random Forest was selected as the best-performing model among the three algorithms.

However, all three models showed relatively low recall, indicating that many actual defaulters were incorrectly classified as non-defaulters.

Therefore, further improvements are required to improve the identification of customers likely to default.
```
